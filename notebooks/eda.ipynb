```python
# Basic libraries
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Settings
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (8,5)
```

```python
df = pd.read_csv("../data/train.csv")
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df.describe()
```

```python
extended_features = [
    'GrLivArea',
    'BedroomAbvGr',
    'FullBath',
    'OverallQual',
    'GarageArea',
    'GarageCars',
    'TotalBsmtSF',
    'YearBuilt',
    'Neighborhood',
    'LotArea',
    'SalePrice'
]
df_core = df[extended_features]
```

```python
df_core.head()
```

GrLivArea:
Above-ground living area in square feet. Represents the total usable living space excluding the basement.

BedroomAbvGr:
Number of bedrooms above ground level.

FullBath:
Number of full bathrooms in the house.

OverallQual:
Overall material and finish quality of the house, rated on a scale from 1 (poor) to 10 (excellent).

GarageArea:
Size of the garage in square feet.

GarageCars:
Garage capacity in terms of number of cars.

TotalBsmtSF:
Total basement area in square feet.

YearBuilt:
Year in which the house was originally constructed.

Neighborhood:
Physical location or neighborhood where the house is situated.

LotArea:
Total lot size (land area) in square feet.

SalePrice:
Final sale price of the house (target variable).

```python
df_core.isnull().sum()
```

```python
#correlation heatmap for numerical features
plt.figure(figsize=(10,6))

sns.heatmap(
    df_core.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```

Correlation analysis indicates that OverallQual and GrLivArea are the strongest predictors of SalePrice, followed by GarageCars, GarageArea, and TotalBsmtSF. 

BedroomAbvGr shows relatively weak correlation with house prices. 

A strong correlation between GarageArea and GarageCars suggests possible multicollinearity, as both features capture similar garage-related information.

we notice strong correlation between garagearea and garagecars, ie multicollinearity.

so we will remove garagearea

```python
df_core = df_core.drop(columns=['GarageArea'])
```

```python
#correlation heatmap for numerical features
plt.figure(figsize=(10,6))

sns.heatmap(
    df_core.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
#neighbourhood vs salesprice
plt.figure(figsize=(14,6))

sns.boxplot(
    x='Neighborhood',
    y='SalePrice',
    data=df_core
)

plt.xticks(rotation=90)
plt.title("Neighborhood vs SalePrice")

plt.show()
```

Neighborhood analysis shows significant variation in house prices across different locations, indicating that location plays a major role in determining property value.

Premium neighborhoods such as NoRidge, NridgHt, and StoneBr exhibit substantially higher median sale prices compared to areas like IDOTRR and MeadowV.

This confirms that neighborhood is an important feature for improving house price prediction.

```python
#overall quality vs salesprice
plt.figure(figsize=(10,6))

sns.boxplot(
    x='OverallQual',
    y='SalePrice',
    data=df_core
)

plt.title("Overall Quality vs SalePrice")

plt.show()
```

there is a clear indication here, higher the quality, better the price

OverallQual shows a very strong positive relationship with SalePrice, indicating that houses with higher construction and material quality tend to have significantly higher prices.

Higher-quality homes also exhibit greater variability in pricing, suggesting the influence of additional premium features in luxury properties.

```python
#year built vs sales price
plt.figure(figsize=(10,6))

sns.scatterplot(
    x='YearBuilt',
    y='SalePrice',
    data=df_core
)

plt.title("Year Built vs SalePrice")

plt.show()
```

YearBuilt shows a moderate positive relationship with SalePrice, indicating that newer houses generally tend to be more expensive.

However, the relationship is not strictly linear, suggesting that additional factors such as location, quality, and renovations also influence property value.

A few older houses still exhibit high prices, likely due to premium locations or renovations.

```python
sns.pairplot(
    df_core[
        [
            'GrLivArea',
            'OverallQual',
            'GarageCars',
            'TotalBsmtSF',
            'SalePrice'
        ]
    ]
)

plt.show()
```

Pairplot analysis confirms strong positive relationships between SalePrice and features such as GrLivArea, OverallQual, GarageCars, and TotalBsmtSF.

The plots also reveal meaningful interactions among housing characteristics, where larger homes generally tend to have higher quality ratings, larger basements, and greater garage capacity.

A few high-value outliers are present but appear to represent genuine premium properties rather than data errors.

```python
#distribution of salesprice
plt.figure(figsize=(8,5))

sns.histplot(
    df_core['SalePrice'],
    kde=True
)

plt.title("Distribution of SalePrice")

plt.show()
```

The SalePrice distribution is right-skewed, with most houses concentrated in the lower-to-mid price range and a smaller number of high-value luxury properties forming a long right tail.

These extreme values appear to represent genuine premium homes rather than erroneous records, and were therefore retained during preprocessing.

```python
#numerical feature distribution
df_core[
    [
        'GrLivArea',
        'OverallQual',
        'GarageCars',
        'TotalBsmtSF',
        'LotArea'
    ]
].hist(
    bins=30,
    figsize=(12,8)
)

plt.suptitle("Feature Distributions")

plt.show()
```

Feature distribution analysis shows that several numerical variables, including GrLivArea, TotalBsmtSF, and LotArea, exhibit right-skewed distributions with a small number of extreme high-value properties.

OverallQual and GarageCars display discrete distributions, with most houses concentrated around average-to-good quality ratings and 2-car garage capacity.

These patterns are consistent with realistic real-estate market characteristics.

```python
#outlier boxplots
plt.figure(figsize=(12,6))

sns.boxplot(
    data=df_core[
        [
            'GrLivArea',
            'TotalBsmtSF',
            'LotArea',
            'SalePrice'
        ]
    ]
)

plt.xticks(rotation=45)

plt.title("Boxplots for Outlier Detection")

plt.show()
```

Boxplot analysis confirms the presence of outliers in features such as LotArea, GrLivArea, TotalBsmtSF, and SalePrice.

However, these extreme values appear to represent genuine large or premium properties rather than data errors, and were therefore retained to preserve the real-world characteristics of the dataset.
